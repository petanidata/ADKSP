## 1) Variabel, tipe data, dan konversi
Variabel adalah tempat menyimpan data. Gunakan `type()` untuk mengecek tipe data.

```python
print("Hello World!")
```

```python
nama = "Andi"
umur = 20
ipk = 3.58
aktif = True

print(nama, umur, ipk, aktif)
print(type(nama), type(umur), type(ipk), type(aktif))

umur_str = "21"
umur_int = int(umur_str)
print(umur_int, type(umur_int))
```

## 2) Operator dasar
Operator dipakai untuk menghitung, membandingkan, dan membuat logika keputusan.

```python
a = 10
b = 3

print("Aritmatika:")
print(a + b, a - b, a * b, a / b, a % b, a ** b)

print("\nPerbandingan:")
print(a > b, a == b, a != b)

print("\nLogika:")
lulus_absen = True
lulus_nilai = False
print("AND ->", lulus_absen and lulus_nilai)
print("OR  ->", lulus_absen or lulus_nilai)
print("NOT ->", not lulus_nilai)
```

## 3) String dasar
String sering dipakai untuk data teks: nama, alamat, komentar, dan lain-lain.

```python
kalimat = "  Belajar Python Dasar  "
print(kalimat)
print(kalimat.strip())
print(kalimat.lower())
print(kalimat.upper())
print(kalimat.replace("Dasar", "Fundamental"))

nama_depan = "Data"
nama_belakang = "Science"
print(f"{nama_depan} {nama_belakang}")
```

## 4) Struktur data utama
Empat struktur dasar yang penting: list, tuple, set, dictionary.

```python
# List (bisa diubah)
nilai = [80, 75, 90]
nilai.append(85)
print("List:", nilai)

# Tuple (tidak bisa diubah)
koordinat = (10, 20)
print("Tuple:", koordinat)

# Set (unik, tanpa urutan tetap)
hobi = {"membaca", "musik", "musik", "lari"}
print("Set:", hobi)

# Dictionary (kunci: nilai)
profil = {"nama": "Sinta", "nim": "2210001", "prodi": "SI"}
profil["angkatan"] = 2022
print("Dictionary:", profil)
print("Nama mahasiswa:", profil["nama"])
```

## 5) Percabangan if-elif-else
Percabangan digunakan untuk memilih aksi berdasarkan kondisi.

```python
nilai_akhir = 78

if nilai_akhir >= 85:
    grade = "A"
elif nilai_akhir >= 70:
    grade = "B"
elif nilai_akhir >= 60:
    grade = "C"
else:
    grade = "D"

print(f"Nilai: {nilai_akhir} -> Grade: {grade}")
```

## 6) Perulangan for dan while
Perulangan dipakai untuk proses berulang, misalnya menampilkan daftar data.

```python
print("for loop 1-5")
for i in range(1, 6):
    if i == 4:
        continue
    print(i)

print("\nwhile loop")
counter = 1
while counter <= 5:
    print("Counter:", counter)
    if counter == 3:
        break
    counter += 1
```

## 7) Fungsi
Fungsi membuat kode lebih rapi, ringkas, dan bisa dipakai ulang.

```python
def hitung_rata_rata(data):
    return sum(data) / len(data)

def sapa(nama, sapaan="Halo"):
    return f"{sapaan}, {nama}!"

daftar_nilai = [70, 80, 90]
print("Rata-rata:", hitung_rata_rata(daftar_nilai))
print(sapa("Budi"))
print(sapa("Alya", sapaan="Selamat pagi"))
```

## 8) Error handling (try-except)
Gunakan `try-except` agar program tidak langsung berhenti saat ada error yang diperkirakan.

```python
angka_str = "10"
pembagi = 0

try:
    hasil = int(angka_str) / pembagi
    print("Hasil:", hasil)
except ZeroDivisionError:
    print("Error: pembagi tidak boleh 0")
except ValueError:
    print("Error: angka_str bukan angka valid")
```

## 9) Mini studi kasus: penilaian mahasiswa
Kita hitung nilai akhir dari tugas, UTS, dan UAS, lalu konversi ke grade.

```python
def hitung_nilai_akhir(tugas, uts, uas):
    return 0.3 * tugas + 0.3 * uts + 0.4 * uas

def konversi_grade(nilai):
    if nilai >= 85:
        return "A"
    if nilai >= 70:
        return "B"
    if nilai >= 60:
        return "C"
    return "D"

data_mahasiswa = [
    {"nama": "Andi", "tugas": 80, "uts": 75, "uas": 88},
    {"nama": "Sinta", "tugas": 90, "uts": 84, "uas": 91},
    {"nama": "Rizky", "tugas": 65, "uts": 70, "uas": 60}
]

for mhs in data_mahasiswa:
    nilai_akhir = hitung_nilai_akhir(mhs["tugas"], mhs["uts"], mhs["uas"])
    grade = konversi_grade(nilai_akhir)
    print(f"{mhs['nama']}: nilai akhir={nilai_akhir:.2f}, grade={grade}")
```
